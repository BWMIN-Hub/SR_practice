## 1. The header

The file is `input.tif`, fetched by the cell above from `dataset/intake/` in this
repository.

It is a THEOS-2 scene over central Bangkok, where the Chao Phraya river runs past
the old city. The product is `ORTHO_PMS`: orthorectified and pan-sharpened, so the
multispectral bands carry the 0.5 m detail of the panchromatic band. Four bands are
stored, in the order red, green, blue, near infrared. The container is a plain
GeoTIFF, which means any GDAL-based reader opens it.

From a terminal, one command prints the whole header and the per-band statistics:

```
gdalinfo -stats input.tif
```

In Python, rasterio is GDAL with a friendlier interface:

```python
import rasterio
with rasterio.open('input.tif') as d:
    d.width, d.height, d.count, d.dtypes, d.crs, d.res
```

Five numbers from that header decide how the file has to be handled. Each one gets
its own look below.

````python
info(PATH)
````

### Size

How many pixels, and how much ground that covers.

````python
show_size(PATH)
````

### Bands

Four of them. The first three are visible light; the fourth is near infrared, which plants reflect strongly and water absorbs.

````python
show_bands(PATH)
````

### Data type

`uint16` is the container. What the sensor put inside it is narrower than that.

````python
show_dtype(PATH)
````

### Coordinate system

Which projection the pixels are pinned to, and where on the Earth the four corners land.

````python
show_crs(PATH)
````

### Pixel size

Zoom in until a single pixel fills the panel, and measure it.

````python
show_pixel(PATH)
````

## 2. The statistics

Do min / max / mean / std fit the bit depth?

````python
stats(PATH, bits=12)
````
