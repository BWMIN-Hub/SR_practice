# Data intake check

Three things to do before a delivered file is used for anything.

```python
%pip install -q rasterio

import sys, urllib.request

BASE = 'https://raw.githubusercontent.com/BWMIN-Hub/SR_practice/main'
urllib.request.urlretrieve(f'{BASE}/lib/intake.py', 'intake.py')
sys.modules.pop('intake', None)
from intake import *

urllib.request.urlretrieve(f'{BASE}/dataset/intake/input.tif', 'input.tif')
```

## 1. The header

Size, band count, dtype, CRS, pixel size.

```python
info('input.tif')
```

## 2. The statistics

Do min / max / mean / std fit the bit depth?

```python
stats('input.tif', bits=12)
```

## 3. The thumbnail

Cloud, empty areas, odd pixels.

```python
preview('input.tif', bits=12)
```
